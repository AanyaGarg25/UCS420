## Q1. Write a unique paragraph (5-6 sentences) about your favorite topic.

```python
import re
import nltk
import matplotlib.pyplot as plt
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.probability import FreqDist

nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)
nltk.download('stopwords', quiet=True)

text = "Technology has changed the way people learn and communicate. Smartphones give us quick access to information. Artificial intelligence helps people solve problems. Online platforms make learning easier for students. New inventions improve healthcare and transportation. I like technology because it creates new ideas and opportunities."

lower_text = text.lower()
clean_text = re.sub(r'[^a-zA-Z0-9\s]', '', lower_text)

print("Lowercase text:")
print(lower_text)
print("Text without punctuation:")
print(clean_text)

words = word_tokenize(clean_text)
sentences = sent_tokenize(text)

print("Word tokens:")
print(words)
print("Sentence tokens:")
print(sentences)

print("Using split():")
print(clean_text.split())
print("Using word_tokenize():")
print(word_tokenize(clean_text))

stop_words = set(stopwords.words('english'))
filtered_words = [word for word in words if word not in stop_words]

print("After stopword removal:")
print(filtered_words)

frequency = FreqDist(filtered_words)
print("Word frequency:")
print(frequency.most_common())

frequency.plot(10, title="Word Frequency")
plt.show()
```

## Q2. Using the same paragraph from Q1:

```python
from nltk.stem import PorterStemmer, WordNetLemmatizer

nltk.download('wordnet', quiet=True)
nltk.download('omw-1.4', quiet=True)

alphabetic_words = re.findall(r'\b[a-zA-Z]+\b', clean_text)
filtered_words = [word for word in alphabetic_words if word not in stop_words]

porter = PorterStemmer()
lemmatizer = WordNetLemmatizer()

stemmed = [porter.stem(word) for word in filtered_words]
lemmatized = [lemmatizer.lemmatize(word) for word in filtered_words]

print("Alphabetic words:")
print(alphabetic_words)
print("After stopword removal:")
print(filtered_words)
print("Stemmed words:")
print(stemmed)
print("Lemmatized words:")
print(lemmatized)

for word, stem, lemma in zip(filtered_words, stemmed, lemmatized):
    print(word, ":", stem, "/", lemma)

print("Stemming cuts words to a root form. Lemmatization gives the meaningful base word. Stemming is faster, while lemmatization is better when correct words are important.")
```

## Q3. Choose 3 short texts of your own.

```python
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
import pandas as pd

texts = [
    "Artificial intelligence helps computers learn from data and recognize images.",
    "Electric vehicles use batteries and electric motors to reduce air pollution.",
    "Online education helps students attend classes and use digital resources."
]

count_vectorizer = CountVectorizer(stop_words='english')
bow = count_vectorizer.fit_transform(texts)
bow_df = pd.DataFrame(bow.toarray(), columns=count_vectorizer.get_feature_names_out())
print("Bag of Words:")
display(bow_df)

tfidf_vectorizer = TfidfVectorizer(stop_words='english')
tfidf = tfidf_vectorizer.fit_transform(texts)
tfidf_df = pd.DataFrame(tfidf.toarray(), columns=tfidf_vectorizer.get_feature_names_out())
print("TF-IDF:")
display(tfidf_df.round(2))

features = tfidf_vectorizer.get_feature_names_out()
for i, row in enumerate(tfidf.toarray()):
    top = row.argsort()[-3:][::-1]
    print("Text", i + 1, "top keywords:")
    print([(features[j], round(row[j], 2)) for j in top])
```

## Q4. Write 2 short texts (4–6 lines each) describing two different technologies.

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

ai_text = """Artificial intelligence helps machines perform tasks.
It can recognize patterns and understand language.
AI is used in healthcare and education.
Machine learning is a part of AI.
AI systems learn from data."""

blockchain_text = """Blockchain stores information in digital blocks.
The blocks are connected in a secure chain.
It works across a distributed network.
Blockchain is used in digital currencies.
It can make records difficult to change."""

def preprocess(text):
    text = text.lower()
    text = re.sub(r'[^a-z\s]', '', text)
    tokens = word_tokenize(text)
    return [word for word in tokens if word not in set(stopwords.words('english'))]

ai_words = preprocess(ai_text)
blockchain_words = preprocess(blockchain_text)

print("AI words:")
print(ai_words)
print("Blockchain words:")
print(blockchain_words)

ai_set = set(ai_words)
blockchain_set = set(blockchain_words)
jaccard = len(ai_set & blockchain_set) / len(ai_set | blockchain_set)
print("Jaccard similarity:", round(jaccard, 3))

vectorizer = TfidfVectorizer(stop_words='english')
matrix = vectorizer.fit_transform([ai_text, blockchain_text])
cosine = cosine_similarity(matrix[0:1], matrix[1:2])[0][0]
print("Cosine similarity:", round(cosine, 3))

print("Jaccard compares common words. Cosine compares TF-IDF values, so it also considers word importance. Cosine is more useful here for comparing the text content.")
```

## Q5. Write a short review for a product or service.

```python
from textblob import TextBlob
from wordcloud import WordCloud

reviews = [
    "The headphones have great sound and a good battery. I am happy with them.",
    "The delivery was late and the product stopped working. The service was poor.",
    "The product is average. It works fine but has ordinary features."
]

results = []
for review in reviews:
    sentiment = TextBlob(review).sentiment
    if sentiment.polarity > 0:
        label = "Positive"
    elif sentiment.polarity < 0:
        label = "Negative"
    else:
        label = "Neutral"

    results.append([review, sentiment.polarity, sentiment.subjectivity, label])

review_df = pd.DataFrame(results, columns=["Review", "Polarity", "Subjectivity", "Sentiment"])
display(review_df)

positive_reviews = [row[0] for row in results if row[3] == "Positive"]
positive_text = " ".join(positive_reviews)

wordcloud = WordCloud(width=800, height=400, background_color="white").generate(positive_text)
plt.figure(figsize=(10, 5))
plt.imshow(wordcloud, interpolation="bilinear")
plt.axis("off")
plt.show()
```

## Q6. Choose your own paragraph (~100 words) as training data.

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense

training_text = "Technology plays an important role in modern life. People use it to communicate, study and work. Computers help us find information quickly. The internet connects people around the world. Artificial intelligence helps machines learn from examples. Mobile applications make daily activities easier. Students use digital tools to understand difficult subjects. Technology also supports hospitals and businesses. It is important to use technology safely and responsibly. New inventions continue to change the world and create opportunities for everyone."

tokenizer = Tokenizer()
tokenizer.fit_on_texts([training_text])
total_words = len(tokenizer.word_index) + 1

sequences = []
for sentence in training_text.lower().split('.'):
    sequence = tokenizer.texts_to_sequences([sentence.strip()])[0]
    for i in range(1, len(sequence)):
        sequences.append(sequence[:i + 1])

max_len = max(len(seq) for seq in sequences)
sequences = pad_sequences(sequences, maxlen=max_len, padding='pre')

X = sequences[:, :-1]
y = tf.keras.utils.to_categorical(sequences[:, -1], num_classes=total_words)

model = Sequential([
    Embedding(total_words, 32),
    LSTM(50),
    Dense(total_words, activation='softmax')
])

model.compile(loss='categorical_crossentropy', optimizer='adam', metrics=['accuracy'])
model.fit(X, y, epochs=100, verbose=0)

def generate_text(seed, next_words=10):
    result = seed.lower()
    for _ in range(next_words):
        seq = tokenizer.texts_to_sequences([result])[0]
        seq = pad_sequences([seq], maxlen=max_len - 1, padding='pre')
        prediction = model.predict(seq, verbose=0)
        index = int(np.argmax(prediction))
        word = next((w for w, i in tokenizer.word_index.items() if i == index), "")
        if not word:
            break
        result += " " + word
    return result.capitalize() + "."

print(generate_text("technology", 10))
print(generate_text("people", 10))
print(generate_text("students", 10))
```
