```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
```

## Q1. Plotting \(M x^2\) and \(M \sin(x)\)

Take a value of `M`, generate values of `x` from -10 to 10, and plot both mathematical functions on the same graph.

```python
M = 2

x = np.linspace(-10, 10, 100)

y1 = M * x**2
y2 = M * np.sin(x)

plt.figure(figsize=(8, 5))
plt.plot(x, y1, linestyle="--", label="y = M*x²")
plt.plot(x, y2, linestyle="-", label="y = M*sin(x)")

plt.title("Mathematical Functions")
plt.xlabel("X")
plt.ylabel("Y")
plt.legend()
plt.grid(True)
plt.show()
```

## Q2. Seaborn Bar Plot of Subject Scores

Create a DataFrame containing subjects and their scores, display the table, and plot the scores using a bar graph.

```python
subjects = ["Maths", "Python", "DBMS", "OS", "AI"]
scores = [85, 90, 78, 88, 95]

df = pd.DataFrame({
    "Subject": subjects,
    "Score": scores
})

df
```

```python
plt.figure(figsize=(8, 5))

ax = sns.barplot(
    x="Subject",
    y="Score",
    data=df,
    hue="Subject",
    palette="Set2",
    legend=False
)

for i, score in enumerate(scores):
    ax.text(i, score + 1, str(score), ha="center")

plt.title("Subject Scores")
plt.xlabel("Subjects")
plt.ylabel("Scores")
plt.grid(axis="y", linestyle="--", alpha=0.7)
plt.show()
```

## Q3. Random Data Visualization

Use Aanya Garg's roll number **1024170172** as the NumPy random seed. Generate 50 random values and create four plots: cumulative sum, scatter plot, line plot, and scatter plot with random noise.

```python
np.random.seed(1024170172)

data = np.random.randn(50)
x = np.arange(1, 51)

cumulative_sum = np.cumsum(data)
noise = np.random.randn(50)

fig, axes = plt.subplots(2, 2, figsize=(12, 8))

axes[0, 0].plot(x, cumulative_sum)
axes[0, 0].set_title("Cumulative Sum")
axes[0, 0].set_xlabel("Index")
axes[0, 0].set_ylabel("Cumulative Sum")
axes[0, 0].grid(True)

axes[0, 1].scatter(x, data)
axes[0, 1].set_title("Random Data Scatter Plot")
axes[0, 1].set_xlabel("Index")
axes[0, 1].set_ylabel("Random Value")
axes[0, 1].grid(True)

axes[1, 0].plot(x, data)
axes[1, 0].set_title("Random Data Line Plot")
axes[1, 0].set_xlabel("Index")
axes[1, 0].set_ylabel("Random Value")
axes[1, 0].grid(True)

axes[1, 1].scatter(x, data + noise)
axes[1, 1].set_title("Scatter Plot with Random Noise")
axes[1, 1].set_xlabel("Index")
axes[1, 1].set_ylabel("Value + Noise")
axes[1, 1].grid(True)

plt.tight_layout()
plt.show()
```

## Q4. Company Sales Data Analysis

Read the monthly company sales dataset from `company_sales_data.csv`. The CSV file is included alongside this notebook.

```python
sales_df = pd.read_csv("company_sales_data.csv")
sales_df
```

### Q4.1. Total Profit of All Months

Plot the total profit for each month using a line graph.

```python
plt.figure(figsize=(10, 6))

sns.lineplot(
    x="month_number",
    y="total_profit",
    data=sales_df,
    marker="o"
)

plt.title("Total Profit of All Months")
plt.xlabel("Month Number")
plt.ylabel("Total Profit")
plt.grid(True)
plt.show()
```

### Q4.2. Monthly Sales of All Products

Plot the monthly sales of face cream, face wash, toothpaste, bathing soap, shampoo, and moisturizer on the same graph.

```python
plt.figure(figsize=(12, 7))

sns.lineplot(x="month_number", y="facecream", data=sales_df, marker="o", label="Face Cream")
sns.lineplot(x="month_number", y="facewash", data=sales_df, marker="o", label="Face Wash")
sns.lineplot(x="month_number", y="toothpaste", data=sales_df, marker="o", label="Toothpaste")
sns.lineplot(x="month_number", y="bathingsoap", data=sales_df, marker="o", label="Bathing Soap")
sns.lineplot(x="month_number", y="shampoo", data=sales_df, marker="o", label="Shampoo")
sns.lineplot(x="month_number", y="moisturizer", data=sales_df, marker="o", label="Moisturizer")

plt.title("Monthly Sales of All Products")
plt.xlabel("Month Number")
plt.ylabel("Units Sold")
plt.legend()
plt.grid(True)
plt.show()
```

### Q4.3. Total Sales of All Products

Calculate the total sales of each product across all months and display the results in a bar graph.

```python
products = [
    "facecream",
    "facewash",
    "toothpaste",
    "bathingsoap",
    "shampoo",
    "moisturizer"
]

total_sales = sales_df[products].sum()

sales_summary = pd.DataFrame({
    "Product": total_sales.index,
    "Total Sales": total_sales.values
})

sales_summary
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    x="Product",
    y="Total Sales",
    data=sales_summary,
    hue="Product",
    palette="Set2",
    legend=False
)

plt.title("Total Sales of All Products")
plt.xlabel("Product")
plt.ylabel("Total Units Sold")
plt.grid(axis="y", linestyle="--", alpha=0.7)
plt.show()
```
